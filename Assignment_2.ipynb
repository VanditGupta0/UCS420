```python
import random

#1
random.seed(1024170307)
L = [random.randint(100, 900) for _ in range(100)]
print("1. Random numbers:", L)

#2
o=[x for x in L if x%2 != 0]
print("2. Odd count:", len(o))
print("Odd numbers:", o)

#3
e=[x for x in L if x%2 ==0]
print("3. Even count:", len(e))
print("Even numbers:", e)

#4
def prime(x):
    if x < 2:
        return False
    for i in range(2, int(x ** 0.5)+1):
        if x % i == 0:
            return False
    return True

p = [x for x in L if prime(x)]
print("4. Prime count:", len(p))
print("Prime numbers:", p)

#5
m = max(L, key=L.count)
print("5. Most frequent:", m)
print("Frequency:", L.count(m))
```

```python
d= [1,0,2,4,1,7,0,3]

A= {x*7 for x in d}
B={x *9 for x in d}

print("A:", A)
print("B:", B)

#6
u=A.union(B)
print("Union:", u)

#7
i=A.intersection(B)
print("Intersection:", i)

#8
ab=A.difference(B)
ba=B.difference(A)

print("A - B:", ab)
print("B - A:", ba)


#9
s=A.symmetric_difference(B)
print("Symmetric difference:", s)

#10
print("A subset of B:", A.issubset(B))
print("B superset of A:", B.issuperset(A))

#11
x=int(input("Enter X: "))
A.discard(x)

print("A after discard:", A)
```

```python
Vandit= np.linspace(10, 100, 25)
print(Vandit)
print("Dimensions:", Vandit.ndim)
print("Shape:", Vandit.shape)
print("Total elements:", Vandit.size)
print("Data type:", Vandit.dtype)
print("Total bytes:", Vandit.nbytes)
trans=Vandit.reshape(25,1)
print(trans)
print(Vandit.T)
```

```python
ucs420_Vandit = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:")
print(ucs420_Vandit)

print("Mean:", np.mean(ucs420_Vandit))
print("Median:", np.median(ucs420_Vandit))
print("Maximum:", np.max(ucs420_Vandit))
print("Minimum:", np.min(ucs420_Vandit))
print("Unique elements:", np.unique(ucs420_Vandit))

reshaped_ucs420_Vandit = ucs420_Vandit.reshape(4, 3)
print("Reshaped Array (4 x 3):")
print(reshaped_ucs420_Vandit)

resized_ucs420_Vandit = np.resize(ucs420_Vandit, (2, 3))
print("Resized Array (2 x 3):")
print(resized_ucs420_Vandit)
```
